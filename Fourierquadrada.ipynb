```python
import numpy as np
import matplotlib.pyplot as plt
```

Série de Fourier Quadrada f(x) = 2a/L f(0) + 2f(0)/pi somatória de n = 1 até infinito 1/n Sen(npia/L) cos (npix/L)

```python
L = 8
x = np.linspace(3*-L, 3*L, 1000)
a = 3
f0 = 5
```

```python
def f(nmax):
    resultado = 2*a*f0/L
    for n in range(1,nmax+1):
        resultado += 2*f0/np.pi*1/n*np.sin(n*np.pi*a/L)*np.cos(n*np.pi*x/L)  
    
    return resultado
```

```python
plt.subplot(2,2,1)
plt.plot(f(2))
plt.subplot(2,2,2)
plt.plot(f(10))
plt.subplot(2,2,3)
plt.plot(f(100))
plt.subplot(2,2,4)
plt.plot(f(1000))
plt.show()
```
